```python
from pulp import *

# Data
foods = ["egg", "bacon", "cereal"]
cost = {"egg": 0.04, "bacon": 0.03, "cereal": 0.02}
vitamin_A = {"egg": 2, "bacon": 4, "cereal": 1}
vitamin_B = {"egg": 3, "bacon": 2, "cereal": 1}

# Problem
prob = LpProblem("Diet_Problem", LpMinimize)

# Decision variables
X = LpVariable.dicts("Food", foods, lowBound=0)

# Objective
prob += lpSum([cost[i] * X[i] for i in foods])

# Constraints
prob += lpSum([vitamin_A[i] * X[i] for i in foods]) >= 16, "Vitamin_A"
prob += lpSum([vitamin_B[i] * X[i] for i in foods]) >= 12, "Vitamin_B"

# Solve and print
prob.solve()
for i in foods:
    print(f"{i}: {X[i].varValue}")
print(f"Cost: ${value(prob.objective):.2f}")
```
