```python
from pulp import *

factories = ['A', 'B', 'C']
markets = ['1', '2', '3', '4']

capacity = {
    'A' : 85,
    'B' : 90, 
    'C' : 110
}

demand = {
    '1' : 55, 
    '2' : 60, 
    '3' : 35, 
    '4' : 50
}

costs = {
    ('A', '1') : 16, ('A', '2') : 13, ('A', '3') : 18, ('A', '4') : 24,
    ('B', '1') : 21, ('B', '2') : 26, ('B', '3') : 23, ('B', '4') : 27,
    ('C', '1') : 24, ('C', '2') : 23, ('C', '3') : 22, ('C', '4') : 29,
}

prob = LpProblem('transportation_problem', LpMinimize)

# dec vars
X = LpVariable.dicts('X', [(i, j) for i in factories for j in markets], lowBound=0, cat='Continuous')

#obj func
prob += lpSum(costs[i, j] * X[i, j] for i in factories for j in markets)

#constraints 
for i in factories: 
    prob += lpSum(X[i, j] for j in markets) <= capacity[i], f'Capacity_{i}'

for j in markets: 
    prob += lpSum(X[i, j] for i in factories) == demand[j], f'Demand_{j}'

prob.solve()

print(f'status: {LpStatus[prob.status]}\ntotal cost: {value(prob.objective)}\noptimal shipments:\n')
for i in factories:
    for j in markets: 
        if X[i, j].varValue > 0:
            print(f'    {i} -> {j} : {X[i, j].varValue}')
```
