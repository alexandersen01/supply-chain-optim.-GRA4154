```python
from pulp import * 

products = ['Clock', 'Radio', 'Toaster']

# params 
c = {
    'Clock' : 7,
    'Radio' : 10, 
    'Toaster' : 5
}

l = {
    'Clock' : 2,
    'Radio' : 3, 
    'Toaster' : 2
}

p = {
    'Clock' : 15,
    'Radio' : 20, 
    'Toaster' : 12
}

d = {
    'Clock' : 200,
    'Radio' : 300, 
    'Toaster' : 150
}

b = 2000
cap = 660

prob = LpProblem('product_mix', LpMaximize)

#dec vars 
X = LpVariable.dicts('X', products, lowBound=0, cat='Integer')

#obj func
prob += lpSum(X[i] * (p[i] - c[i]) for i in products), 'Total_Profit'

#constraints 
for i in products:
    prob += X[i] <= d[i], f'Demand_{i}'

prob += lpSum(X[i] * c[i] for i in products) <= b, 'Budget'

prob += lpSum(X[i] * l[i] for i in products) <= cap, 'Labor'

prob.solve()

print(f'status: {LpStatus[prob.status]}\noptimal production (X_i):\n')
for i in products: 
    print(f'    {i} : {X[i].varValue}')
print(f'\nmax profit: {value(prob.objective)}')
```
