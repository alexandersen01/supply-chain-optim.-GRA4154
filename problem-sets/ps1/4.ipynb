```python
from pulp import *

sources = [1, 2, 3]
destinations = [1, 2, 3]

costs = {
    (1, 1) : 3, (1, 2) : 12, (1, 3) : 8,
    (2, 1) : 10, (2, 2) : 5, (2, 3) : 6,
    (3, 1) : 6, (3, 2) : 7, (3, 3) : 10,
}

supply = {1 : 90, 2 : 30, 3 : 100}

capacity = {1 : 70, 2 : 110, 3 : 80}

prob = LpProblem('transportation', LpMinimize)

x = LpVariable.dicts('x', [(i, j) for i in sources for j in destinations], lowBound=0, cat='Continuous')

prob += lpSum(costs[i, j] * x[i, j] for i in sources for j in destinations)

for i in sources: 
    prob += lpSum(x[i, j] for j in destinations) == supply[i], f'Supply_{i}'

for j in destinations:
    prob += lpSum(x[i, j] for i in sources) <= capacity[j], f'Capacity_{j}'

prob.solve()

print(f'status: {LpStatus[prob.status]}\nminimum cost = {value(prob.objective)}\noptimal shipments:')
for i in sources:
    for j in destinations: 
        print(f'    x_{i}{j} = {x[i, j].varValue}')
```
